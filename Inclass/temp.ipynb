```python
# Plotting overall D-Cache miss rates across cache sizes and associativities
import matplotlib.pyplot as plt

# Data organized by associativity
data = {
    1: {
        "sizes": [1, 2, 4, 8],
        "miss_rates": [8.0265, 5.4626, 4.3207, 3.4520]
    },
    4: {
        "sizes": [1, 2, 4, 8],
        "miss_rates": [4.2999, 3.3947, 3.0497, 2.8802]
    },
    8: {
        "sizes": [1, 2, 4, 8],
        "miss_rates": [4.2302, 3.3421, 3.0110, 2.8748]
    },
    3: {
        "sizes": [6, 12],
        "miss_rates": [2.9630, 2.8292]
    }
}

plt.figure()

for assoc, values in data.items():
    plt.plot(values["sizes"], values["miss_rates"], marker='o', label=f"{assoc}-way")

plt.xlabel("L1D Cache Size (KB)")
plt.ylabel("D-Cache Miss Rate (%)")
plt.title("D-Cache Miss Rate vs Cache Size for Different Associativities")
plt.legend()
plt.grid(True)
plt.savefig('Cache.png', dpi=300)
plt.show()
```
