# AML s26 - Cross-Validation (CV) and Leave-One-Out (LOO) CV

### Import Libraries

```python
import numpy as np
import pandas as pd

from sklearn.linear_model import LinearRegression
from sklearn.model_selection import cross_val_score, KFold
```

### Load data

```python
df = pd.read_csv('data_ridge_lasso.csv')

X = df.drop('Y', axis=1).values
y = df['Y'].values
```

### CV using all data 

```python
# Create regression model instance 
model = LinearRegression()
k = 5

# Set up K-Fold cross-validation
kfold = KFold(n_splits=k, shuffle=True, random_state=22)

# Perform cross-validation with MSE metric
cv_mse = cross_val_score(model, X, y, cv=kfold, scoring='neg_mean_squared_error')  

# Convert to positive values of MSE 
cv_mse = -cv_mse
```

### Show CV results 

```python
print(f"\nCV Results ({k} folds):")
print("-" * 25)
print(f"   {'Fold':<10} {'MSE':<12}")
print("-" * 25)

for i in range(k):
    print(f"   Fold {i+1:<5} {cv_mse[i]:<12.4f}")

print("-" * 25)
print(f"   {'Mean':<10} {cv_mse.mean():<12.4f}")
print(f"   {'Std':<10} {cv_mse.std():<12.4f}")
```

### Now use Leave-One-Out (LOO) CV

```python
from sklearn.model_selection import LeaveOneOut

loo_mse = cross_val_score(model, X, y, cv=LeaveOneOut(), scoring='neg_mean_squared_error') 
# Note: R^2 is not meaningful when using LOO approach! 

loo_mse = -loo_mse

np.set_printoptions(precision=3)
print(f"MSE: ", loo_mse)

print(f"\nLOO Result:")
print("-" * 25)
print(f"   {'Mean':<10} {loo_mse.mean():<12.4f}")
print(f"   {'Std':<10} {loo_mse.std():<12.4f}")
```
