```python
import pandas as pd
import numpy as np

def generate_synthetic_data(rows_per_class=100):
    np.random.seed(42)  # For reproducibility
    data = []
    
    # Define profiles for 11 classes (Target mean for specific feature groups)
    # Profiles: (X1-X6_scale, X7-X12_mean, X19-X24_mean, X37-X42_mean, X43-X48_mean)
    profiles = {
        1:  (1e-5,  0.03,  0.9,   10.0,  -1.50),
        2:  (1e-4,  0.08,  1.4,   35.0,  -1.31),
        3:  (1e-6, -0.09,  0.18, -15.0,  -1.65),
        4:  (1e-4,  0.15,  2.2,   70.0,  -1.13),
        5:  (1e-6,  0.01,  0.5,    2.5,  -1.50),
        6:  (1e-4,  0.55,  0.01, -85.0,  -1.98),
        7:  (1e-5, -0.04,  0.4,   18.0,  -1.52),
        8:  (1e-6,  0.12,  1.8,  -42.0,  -1.41),
        9:  (1e-4,  0.0004,0.9,    5.5,  -1.49),
        10: (1e-6, -0.12,  0.12,  45.0,  -1.58),
        11: (1e-7,  0.0,   1.0,    0.0,  -1.50)
    }

    for class_id, (prec, mid_m, ratio_m, high_m, base_m) in profiles.items():
        for _ in range(rows_per_class):
            row = {}
            
            # X1 - X6: High precision micro-fluctuations
            for i in range(1, 7):
                row[f'X{i}'] = np.random.normal(0, prec)
            
            # X7 - X12: Mid-range fluctuations
            for i in range(7, 13):
                row[f'X{i}'] = np.random.normal(mid_m, abs(mid_m) * 0.1 + 0.01)
            
            # X13 - X18: Transition variables (small noise)
            for i in range(13, 19):
                row[f'X{i}'] = np.random.uniform(0, 0.005)
            
            # X19 - X24: Ratios/Feedback loops
            for i in range(19, 25):
                row[f'X{i}'] = np.random.normal(ratio_m, 0.05)
            
            # X25 - X36: Intermediate variance
            for i in range(25, 37):
                row[f'X{i}'] = np.random.normal(mid_m * 2, 0.1)
            
            # X37 - X42: High magnitude sensor data
            for i in range(37, 43):
                row[f'X{i}'] = np.random.normal(high_m, abs(high_m) * 0.05 + 1.0)
            
            # X43 - X48: Stable environmental baseline
            for i in range(43, 49):
                row[f'X{i}'] = np.random.normal(base_m, 0.005)
            
            row['Y'] = class_id
            data.append(row)

    return pd.DataFrame(data)

# Generate the dataset
df = generate_synthetic_data(rows_per_class=100)

# Shuffle the dataset so classes aren't in perfect blocks
df = df.sample(frac=1).reset_index(drop=True)

# Save to CSV
df.to_csv('synthetic_11_class_data.csv', index=False)

print(f"Success! Generated {len(df)} rows across 11 classes.")
print(df.head())
```
